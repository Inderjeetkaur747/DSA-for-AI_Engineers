```python
# Problem 1 – Find Maximum Element
# Given an array, return maximum element.
def find_max(arr):
  if not arr:
    return None
  max_ele = arr[0]
  for num in range(1, len(arr)):
    if arr[num] > max_ele:
      max_ele = arr[num]
  return max_ele

arr = [1,2,4,3,6,9,8]
print(find_max(arr))
```

```python
# Problem 2 – Second Largest Element

# Given array, return second largest number.

# Conditions:

# Do not sort

# One pass preferred

# Handle duplicates
#Two Variable Tracking Pattern
def second_largest(arr):
    if len(arr) < 2:
        return None

    first = float('-inf')
    second = float('-inf')

    for num in arr:
        if num > first:
            second = first
            first = num
        elif first > num > second:
            second = num

    if second == float('-inf'):
        return None

    return second

arr = [1,2,4,3,6,9,-9,-8,-1]
print(second_largest(arr))


```

```python
# Problem 3 – Move Zeros
# Given an integer array nums, move all 0's to the end
# while maintaining the relative order of non-zero elements.
# Do it in-place.
#Two Pointer Compaction Pattern

def move_zeros(arr):
  insert_arr = 0
  for num in range(len(arr)):
    if arr[num]!=0:
      arr[insert_arr] = arr[num]
      insert_arr+=1

  for num in range(insert_arr, len(arr)):
    arr[num]=0
  return arr

arr = [1,3,0,4,0,2,0,5]
print(move_zeros(arr))
```

```python
# Problem 4 – Remove Element (LeetCode 27)
# Given array nums and a value val,
# remove all occurrences of val in-place.
# Return the new length.
def remove_element(arr,val):
  write_index = 0
  for i in range(len(arr)):
    if arr[i]!=val:
      arr[write_index] = arr[i]
      write_index+=1
  return write_index

arr = [1,2,2,3,4]
val = 2
print(remove_element(arr,val))
```

```python
# Problem 5 – Best Time to Buy and Sell Stock (LeetCode 121)
# Given:
# prices = [7,1,5,3,6,4]
# Return maximum profit.
# Rules:
# You must buy before selling.
# Only one transaction allowed.
# Example:
# Buy at 1
# Sell at 6
# Profit = 5
def max_profit(prices):
  if not prices:
    return 0

  min_price = prices[0]
  max_profit = 0
  for i in range(1, len(prices)):
    if prices[i] < min_price:
      min_price = prices[i]
    else:
      profit = prices[i] - min_price
      max_profit = max(max_profit,profit)

  return max_profit


prices = [7,1,5,3,6,4]
print(max_profit(prices))


#Running Minimum + Profit Tracking Pattern

# Very common in:

# Stock problems

# Subarray problems

# Maximum difference problems
```

```python
# Problem 6 – Contains Duplicate (LeetCode 217)
# Given an array, return True if any value appears at least twice.
# Example:
# [1,2,3,1] → True
# [1,2,3,4] → False
# ⚠ Important:
# Do NOT use sorting.
# Do NOT use built-in set.
# Think pure array logic first.
def contains_duplicate(val):
  n = len(val)
  for i in range(n):
    for j in range(i+1, n):
      if val[i]==val[j]:
        return True

  return False


val = [1,2,3,1]
print(contains_duplicate(val))

# Outer loop = n
# Inner loop = n

# Total = n²

# Time Complexity = O(n²)
# Space Complexity = O(1)
# This problem shows:
# Arrays alone are inefficient for lookup problems.
# We need faster lookup.
# And that’s exactly why Hashing exists.
#Arrays are bad at searching quickly.
```

```python
def contains_duplicate(arr):
  seen = set()
  for num in arr:
    if num in seen:
      return True
    seen.add(num)
  return False

arr = [1,2,3,1]
print(contains_duplicate(arr))
```

```python
# Problem – Two Sum (LeetCode 1)
# Given:
# nums = [2,7,11,15]
# target = 9
# Return indices of two numbers that add to target.
def Two_sum(num, target):
  hash_map = {}
  for i, n in enumerate(num):
    complement = target - n
    if complement in hash_map:
      return [hash_map[complement],i]
    hash_map[n] = i
  return []

num = [2,7,11,15]
target = 9
print(Two_sum(num, target))

```

```python
# Next problem:
# Valid Anagram (LeetCode 242)
#frequency counting pattern
def is_anagram(s, t):

    if len(s) != len(t):
        return False

    freq = {}

    # Count characters in s
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1 #ch = 'a': freq.get('a', 0) is 0. freq['a'] = 0 + 1. freq is {'a': 1}.

    # Subtract using t
    for ch in t:
        if ch not in freq:
            return False

        freq[ch] -= 1

        if freq[ch] < 0:
            return False

    return True

s = "anagram"
t = "nagaram"
print(is_anagram(s,t))
```

```python
# Frequency Counting Level 2
# Problem – Majority Element (LeetCode 169)
def majority_element(arr):
  if not arr:
    return None
  freq = {}
  for n in arr:
    freq[n]= freq.get(n, 0)+1

    if freq[n]>len(arr)//2:
      return n

arr = [2,2,1,1,1,2,2]
print(majority_element(arr))


```

```python
#Majority Element – Optimal Solution
#Boyer-Moore Voting Algorithm
# Time = O(n)
# Space = O(1)
def majority_element(arr):
  count = 0
  candidate = None
  for num in arr:
    if count == 0:
      candidate = num

    if num == candidate:
      count+=1
    else:
      count-=1

  return candidate

arr = [2,2,1,1,2,2,1]
print(majority_element(arr))


#The majority element appears more than n/2 times, so even if we pair each occurrence of
#the majority element with a different element and cancel them out,
#the majority element will still remain.
```

```python
#Frequency Counting Advanced
#Top K Frequent Elements (LeetCode 347)
#sorting
def fre_count(arr, k):
  if not arr:
    return None

  freq = {}

  for num in arr:
    freq[num] = freq.get(num,0)+1

  sorted_nums = sorted(freq, key = freq.get, reverse = True)
  return sorted_nums[:k]

arr = [1,2,3,2,3,2,3,2,2,3]
k=2
print(fre_count(arr,k))


```

```python
#using heap
import heapq
def fre_ele(arr, k):
  if not arr:
    return None
  freq = {}
  for num in arr:
    freq[num] = freq.get(num, 0)+1

  return heapq.nlargest(k,freq, key = freq.get)

arr =[1,2,3,2,3,2,3,2,2,3]
print(fre_ele(arr,2))

#heap is faster when k is small because heap complexity is O(n log k) and sorting complexit is O(n log n).
#since k is much smaller than n. log k is smaller than log n, making heap more efficient.

```

```python
#Prefix Sum- Store running sums from the beginning
#Prefix Sum Problem 1
# Write function:
# Return prefix sum array
def prefix_sum(arr):
  prefix = [0]*len(arr)
  prefix[0] = arr[0]
  for i in range(1, len(arr)):
    prefix[i] = prefix[i-1]+arr[i]
  return prefix

arr = [1,2,3,4]
print(prefix_sum(arr))


#How To Recognize Prefix Sum Problem
#If question says:
# "subarray sum"
# "range sum"
# "continuous sum"
# "sum equals k"
#Think Prefix Sum.
```

```python
def range_sum(arr, left, right):
  prefix = [0]*len(arr)
  prefix[0] = arr[0]
  for i in range(1, len(arr)):
    prefix[i] = prefix[i-1]+arr[i]

  if left==0:
    return prefix[right]
  return prefix[right]- prefix[left-1]

arr = [1,2,3,4,5]
print(range_sum(arr,0,4))
```

```python
def subarray(arr,k):
  current_sum = 0
  count = 0
  prefix_map = {0:1}
  for num in arr:
    current_sum+=num

    if current_sum - k in prefix_map:
      count+=prefix_map[current_sum-k]
    prefix_map[current_sum]= prefix_map.get(current_sum,0)+1

  return count

arr = [1,2,3]
k = 3
print(subarray(arr,k))


```
